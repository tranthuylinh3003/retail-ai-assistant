```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python

df = pd.read_csv("../data/online_retail_II.csv", parse_dates=["InvoiceDate"])
df = df[(df["Quantity"] > 0) & (df["Price"] > 0)].copy()
df["Revenue"] = df["Quantity"] * df["Price"]
df["YearMonth"] = df["InvoiceDate"].dt.to_period("M")

# keep real products only: real StockCodes start with a digit (excludes POST, DOT, M, AMAZONFEE...)
products = df[df["StockCode"].astype(str).str.match(r"^\d")]
```

```python
top10 = (
    products.groupby("Description")["Revenue"].sum()
    .nlargest(10)
    .sort_values()          # smallest first, so the biggest ends up on top
)

fig, ax = plt.subplots(figsize=(10, 5))
ax.barh(top10.index, top10.values)
ax.set_title("Top 10 products by revenue")
ax.set_xlabel("Revenue (GBP)")
ax.xaxis.set_major_formatter(lambda x, _: f"{x/1000:,.0f}K")
plt.tight_layout()
plt.show()
```

```python
top5 = df.groupby("Week")["Revenue"].sum().nlargest(5)

fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(top5.index, top5.values)
ax.set_title("Top 5 weekly by revenue")
ax.set_ylabel("Revenue (GBP)")
plt.tight_layout()
plt.show()
```

```python
top5 = df.groupby("Hour")["Revenue"].sum().nlargest(5)

fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(top5.index, top5.values)
ax.set_title("Top 5 Hour by revenue")
ax.set_ylabel("Revenue (GBP)")
plt.tight_layout()
plt.show()
```
