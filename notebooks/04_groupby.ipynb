```python
import pandas as pd

df = pd.read_csv("../data/online_retail_II.csv", parse_dates=["InvoiceDate"])
```

```python
df["Revenue"] = df["Quantity"] * df["Price"]
```

```python
total_rev_country = df.groupby("Country")["Revenue"].sum().sort_values(ascending=False)
total_rev_country.head(5)
```

```python
uk_share = total_rev_country["United Kingdom"] / total_rev_country.sum()
print(uk_share * 100)
```

```python
top_10_product =df.groupby("Description")["Revenue"].sum().sort_values(ascending=False)
top_10_product.head(10)
```

```python
top_10_product_unit =df.groupby("Description")["Quantity"].sum().sort_values(ascending=False)
top_10_product_unit.head(10)
```

```python
customer_unique = df.groupby("Country")["Customer ID"].nunique().sort_values(ascending=False)
customer_unique.head(10)
```

```python
inv_rev = df.groupby("Invoice")["Revenue"].sum()
print(inv_rev.mean())
print(inv_rev.median())
```

```python
inv_per_cust = df.groupby("Customer ID")["Invoice"].nunique()
print(inv_per_cust.min(), inv_per_cust.median(), inv_per_cust.max())
print((inv_per_cust == 1).mean() * 100)   # % of one-time buyers
```

```python
cust_rev = df.groupby("Customer ID")["Revenue"].sum().sort_values(ascending=False)
top10 = cust_rev.head(10)
print(top10)

share = top10.sum() / cust_rev.sum() * 100
print(share)
```
