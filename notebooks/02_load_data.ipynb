```python
import pandas as pd

sheets = pd.read_excel("../data/online_retail_II.xlsx", sheet_name=None)
print(sheets.keys())
```

```python
# stack the two sheets into one table (like UNION ALL in SQL)
df = pd.concat(sheets.values(), ignore_index=True)
```

```python
# save a CSV copy so next time loading takes seconds
df.to_csv("../data/online_retail_II.csv", index=False)
```
